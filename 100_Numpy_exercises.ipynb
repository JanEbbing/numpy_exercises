#### 17. What is the result of the following expression? (★☆☆)
```python
0 * np.nan
np.nan == np.nan
np.inf > np.nan
np.nan - np.nan
np.nan in set([np.nan])
0.3 == 3 * 0.1
```

````python
print(0 * np.nan) # arguable what this should return, as 0 * any number stays 0 (I guess nan is not a number :) ), but nan usually propagates. In this case, result is np.nan
print(np.nan == np.nan) # nan is not equal to anything, so False
print(np.inf > np.nan) # False, all these boolean binary operators with NaN return False. All boolean operators with NaN will return NaN
print(np.nan - np.nan) # np.nan, see above
print(np.nan in set([np.nan])) # True, np.nan gets hashed to the same value. Python then checks reference equality with `is`, which evaluates to `True` for np.nan,
# otherwise it would evaluate equality with `__eq__()`, which would be False (see above)
# Note that this is technically a collision in a hash table/hash set, as the elements are not equal to each other
# There is more weirdness around this, e.g. set([np.nan] * 5) will contain one element, whereas set([np.nan for _ in range(5)]) will contain 5 elements
print(0.3 == 3 * 0.1)  # False, floating point imprecision. Use abs(float1 - float) < EPSILON for a small EPSILON, e.g. 0.00001 for equality of floats
# Note that this being true or false depends somewhat on the number you choose, e.g. 0.5 == 5 * 0.1 might evaluate to true as 0.5 can be expressed
# perfectly in IEEE 754 (0.1 cannot though, this is still `True` on my laptop)
````

#### 27. Consider an integer vector Z, which of these expressions are legal? (★☆☆)
```python
Z**Z
2 << Z >> 2
Z <- Z
1j*Z
Z/1/1
Z<Z>Z
```

````python
Z = np.random.randint(0, 5, 5)
print(Z)
print("\n")

print(Z ** Z) # works, element wise exponentiation. Note that 0^0 = 1
print(2 << Z >> 2) # works, element wise bit shift, first to the left, then the right. Note that this in effect rounds to the nearest power of 2, you don't get the original value
print(Z <-Z) # works, gives a numpy array of the same dimension as Z full of `False` (element wise comparison; no number is smaller than the negative of itself)
print(1j * Z) # works, 1j is the imaginary number in python, multiplication is element-wise, so this gives an array of the same dimension as the input, with each value being the complex number 0 + entry * i
print(Z/1/1) # works, just gives Z with the entries being floats (element wise division, then again element wise division)
# print(Z < Z > Z) # In python a < b > c resolves to a < b and b > c, however for numpy arrays `and` is not defined, you'd need to use `np.logical_and`, so this throws at runtime
````

#### 41. How to sum a small array faster than np.sum? (★★☆)

````python
import timeit

to_sum = np.array([1,2,3,4,5,6,7,8,-10])

# Also swap order of timeit operations to ensure there are no cache effects
print("np.add.reduce()")
print(timeit.timeit("np.add.reduce(np.array([1,2,3,4,5,6,7,8,-10]))", setup="import numpy as np"))

print("np.sum()")
print(timeit.timeit("np.sum(np.array([1,2,3,4,5,6,7,8,-10]))", setup="import numpy as np"))


print("Correctness:")
print(np.sum(to_sum))
print(np.add.reduce(to_sum))
````
